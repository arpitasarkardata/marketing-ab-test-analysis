```python
import pandas as pd
from statsmodels.stats.proportion import proportions_ztest

df = pd.read_csv('marketing_AB.csv')

# Convert True/False text into 1/0 numbers
df['converted'] = df['converted'].astype(str).map({'True': 1, 'False': 0})

ad_group = df[df['test group'] == 'ad']['converted']
psa_group = df[df['test group'] == 'psa']['converted']

count = [ad_group.sum(), psa_group.sum()]
nobs = [len(ad_group), len(psa_group)]
z_stat, p_value = proportions_ztest(count, nobs)

ad_rate = ad_group.mean()
psa_rate = psa_group.mean()
lift = ad_rate - psa_rate

print(f"Ad group conversion rate: {ad_rate:.4f}")
print(f"PSA group conversion rate: {psa_rate:.4f}")
print(f"Absolute lift: {lift:.4f}")
print(f"z-statistic: {z_stat:.3f}")
print(f"p-value: {p_value:.5f}")
```
